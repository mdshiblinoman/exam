```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read and convert image
image = cv2.imread("cat.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Spatial Resolution - create resized versions
original = image

low_spatial = cv2.resize(image, (100, 100))
medium_spatial = cv2.resize(image, (300, 300))
high_spatial = cv2.resize(image, (600, 600))
```

```python
# Display Spatial Resolution
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(low_spatial)
plt.title("Low Spatial Resolution (100x100)")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(medium_spatial)
plt.title("Medium Spatial Resolution (300x300)")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(high_spatial)
plt.title("High Spatial Resolution (600x600)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Intensity Resolution - process different intensity levels
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

def reduce_intensity(image, levels):
    factor = 256 // levels
    return (image // factor) * factor

intensity_256 = gray
intensity_16 = reduce_intensity(gray, 16)
intensity_4 = reduce_intensity(gray, 4)
intensity_2 = reduce_intensity(gray, 2)
```

```python
# Display Intensity Resolution
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(intensity_256, cmap="gray")
plt.title("256 Intensity Levels")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(intensity_16, cmap="gray")
plt.title("16 Intensity Levels")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(intensity_2, cmap="gray")
plt.title("2 Intensity Levels")
plt.axis("off")

plt.tight_layout()
plt.show()
```
