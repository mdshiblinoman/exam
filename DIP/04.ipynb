## Problem 4
Write a Python program to extract color channels (Red, Green, Blue) from an RGB image.


```python
import matplotlib.pyplot as plt
import numpy as np

image = plt.imread("cat.png")
height, width = image.shape[:2]

# Extract RGB channels manually using array indexing.
red = np.zeros((height, width), dtype=image.dtype)
green = np.zeros((height, width), dtype=image.dtype)
blue = np.zeros((height, width), dtype=image.dtype)

for row in range(height):
    for col in range(width):
        red[row, col] = image[row, col, 0]
        green[row, col] = image[row, col, 1]
        blue[row, col] = image[row, col, 2]

# Display each channel separately.
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(red, cmap="Reds")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(green, cmap="Greens")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(blue, cmap="Blues")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```
