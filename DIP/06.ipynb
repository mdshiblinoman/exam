```python
import cv2
import matplotlib.pyplot as plt

# Read Two Images

image1 = cv2.imread("cat.png")
image2 = cv2.imread("Rose1.png")

# Resize image2 to the same size as image1
image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

# 1. Arithmetic Operations

# Addition
addition = cv2.add(image1, image2)

# Subtraction
subtraction = cv2.subtract(image1, image2)

# 2. Logical Operations

# AND operation
logical_and = cv2.bitwise_and(image1, image2)

# OR operation
logical_or = cv2.bitwise_or(image1, image2)

# NOT operation
logical_not = cv2.bitwise_not(image1)

# Convert BGR to RGB for displaying

image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)

addition_rgb = cv2.cvtColor(addition, cv2.COLOR_BGR2RGB)
subtraction_rgb = cv2.cvtColor(subtraction, cv2.COLOR_BGR2RGB)
logical_and_rgb = cv2.cvtColor(logical_and, cv2.COLOR_BGR2RGB)
logical_or_rgb = cv2.cvtColor(logical_or, cv2.COLOR_BGR2RGB)
logical_not_rgb = cv2.cvtColor(logical_not, cv2.COLOR_BGR2RGB)

# Display Results

plt.figure(figsize=(12, 10))

plt.subplot(2, 4, 1)
plt.imshow(image1_rgb)
plt.title("Image 1")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(image2_rgb)
plt.title("Image 2")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(addition_rgb)
plt.title("Addition")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(subtraction_rgb)
plt.title("Subtraction")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(logical_and_rgb)
plt.title("AND")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(logical_or_rgb)
plt.title("OR")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(logical_not_rgb)
plt.title("NOT (Image 1)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# 1. MANUAL: ARITHMETIC OPERATIONS

# Manual Addition (with clipping to prevent overflow)
# Result should be clipped to [0, 255]
addition_manual = np.minimum(image1.astype(np.float32) + image2.astype(np.float32), 255).astype(np.uint8)


# Manual Subtraction (with clipping to prevent underflow)
# Result should be clipped to [0, 255]
subtraction_manual = np.maximum(image1.astype(np.float32) - image2.astype(np.float32), 0).astype(np.uint8)


# 2. MANUAL: LOGICAL OPERATIONS

# Manual AND operation (bitwise AND)
logical_and_manual = np.bitwise_and(image1, image2)

# Manual OR operation (bitwise OR)
logical_or_manual = np.bitwise_or(image1, image2)

# Manual NOT operation (bitwise NOT)
logical_not_manual = np.bitwise_not(image1)

# 3. MANUAL: COLOR CONVERSION (BGR to RGB)

# Manual BGR to RGB conversion
# OpenCV loads as BGR, we need to swap channels manually
B1 = image1[:, :, 0]
G1 = image1[:, :, 1]
R1 = image1[:, :, 2]
image1_rgb_manual = np.stack([R1, G1, B1], axis=2)

B2 = image2[:, :, 0]
G2 = image2[:, :, 1]
R2 = image2[:, :, 2]
image2_rgb_manual = np.stack([R2, G2, B2], axis=2)

# Convert operation results
B_add = addition_manual[:, :, 0]
G_add = addition_manual[:, :, 1]
R_add = addition_manual[:, :, 2]
addition_rgb_manual = np.stack([R_add, G_add, B_add], axis=2)

B_sub = subtraction_manual[:, :, 0]
G_sub = subtraction_manual[:, :, 1]
R_sub = subtraction_manual[:, :, 2]
subtraction_rgb_manual = np.stack([R_sub, G_sub, B_sub], axis=2)

B_and = logical_and_manual[:, :, 0]
G_and = logical_and_manual[:, :, 1]
R_and = logical_and_manual[:, :, 2]
logical_and_rgb_manual = np.stack([R_and, G_and, B_and], axis=2)

B_or = logical_or_manual[:, :, 0]
G_or = logical_or_manual[:, :, 1]
R_or = logical_or_manual[:, :, 2]
logical_or_rgb_manual = np.stack([R_or, G_or, B_or], axis=2)

B_not = logical_not_manual[:, :, 0]
G_not = logical_not_manual[:, :, 1]
R_not = logical_not_manual[:, :, 2]
logical_not_rgb_manual = np.stack([R_not, G_not, B_not], axis=2)

# DISPLAY RESULTS

plt.figure(figsize=(14, 10))

plt.subplot(2, 4, 1)
plt.imshow(image1_rgb_manual)
plt.title("Image 1\n(Manual BGR→RGB)", fontweight='bold')
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(image2_rgb_manual)
plt.title("Image 2\n(Manual BGR→RGB)", fontweight='bold')
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(addition_rgb_manual)
plt.title("Addition\n(img1 + img2)", fontweight='bold')
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(subtraction_rgb_manual)
plt.title("Subtraction\n(img1 - img2)", fontweight='bold')
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(logical_and_rgb_manual)
plt.title("AND\n(img1 & img2)", fontweight='bold')
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(logical_or_rgb_manual)
plt.title("OR\n(img1 | img2)", fontweight='bold')
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(logical_not_rgb_manual)
plt.title("NOT\n(~img1)", fontweight='bold')
plt.axis("off")

plt.tight_layout()
plt.show()
```
