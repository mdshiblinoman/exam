## Problem 6
Perform addition, subtraction, AND, OR and NOT operations on two images.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image1 = cv2.imread("cat.png")
image2 = cv2.imread("Rose1.png")

# Use the same common size for both images.
height = image1.shape[0]
if image2.shape[0] < height:
    height = image2.shape[0]
width = image1.shape[1]
if image2.shape[1] < width:
    width = image2.shape[1]
image1 = image1[:height, :width]
image2 = image2[:height, :width]

addition = np.zeros_like(image1)
subtraction = np.zeros_like(image1)
and_image = np.zeros_like(image1)
or_image = np.zeros_like(image1)
not_image = np.zeros_like(image1)

for row in range(height):
    for col in range(width):
        for channel in range(3):
            a = int(image1[row, col, channel])
            b = int(image2[row, col, channel])

            total = a + b
            addition[row, col, channel] = 255 if total > 255 else total
            difference = a - b
            subtraction[row, col, channel] = 0 if difference < 0 else difference
            and_image[row, col, channel] = a & b
            or_image[row, col, channel] = a | b
            not_image[row, col, channel] = 255 - a

images = [image1, image2, addition, subtraction, and_image, or_image, not_image]
titles = ["Image 1", "Image 2", "Addition", "Subtraction", "AND", "OR", "NOT"]

plt.figure(figsize=(14, 8))
for i in range(7):
    plt.subplot(2, 4, i + 1)
    plt.imshow(images[i][:, :, ::-1])
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
