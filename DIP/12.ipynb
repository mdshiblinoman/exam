## Problem 12
Write a Python program to implement spatial domain smoothing/averaging filters.


```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Read Image
image = cv2.imread("cat.png", 0)

# Spatial Domain Smoothing Filters (Manual Implementation)
def apply_averaging_filter(image, kernel_size):
    h, w = image.shape
    pad = kernel_size // 2
    
    # Create output image
    output = np.zeros_like(image, dtype=np.float32)
    
    # Add padding to the image
    padded = np.pad(image, pad, mode='reflect')
    
    # Apply filter
    for i in range(h):
        for j in range(w):
            # Extract the region
            region = padded[i:i+kernel_size, j:j+kernel_size]
            # Calculate average
            output[i, j] = np.mean(region)
    
    return np.uint8(output)

# 3x3 Averaging Filter
smooth_3x3 = apply_averaging_filter(image, 3)

# 5x5 Averaging Filter
smooth_5x5 = apply_averaging_filter(image, 5)

# 7x7 Averaging Filter
smooth_7x7 = apply_averaging_filter(image, 7)

# Display Results
plt.figure(figsize=(12, 8))

# Original
plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

# 3x3
plt.subplot(2, 2, 2)
plt.imshow(smooth_3x3, cmap="gray")
plt.title("3 × 3 Averaging Filter")
plt.axis("off")

# 5x5
plt.subplot(2, 2, 3)
plt.imshow(smooth_5x5, cmap="gray")
plt.title("5 × 5 Averaging Filter")
plt.axis("off")

# 7x7
plt.subplot(2, 2, 4)
plt.imshow(smooth_7x7, cmap="gray")
plt.title("7 × 7 Averaging Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
