```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# Load the image
image = cv2.imread("cat.png")

# OpenCV loads as BGR, we need RGB
B = image[:, :, 0]
G = image[:, :, 1]
R = image[:, :, 2]
rgb_image = np.stack([R, G, B], axis=2)


# Manual grayscale conversion using weighted formula
# Gray = 0.299*R + 0.587*G + 0.114*B
R_channel = rgb_image[:, :, 0]
G_channel = rgb_image[:, :, 1]
B_channel = rgb_image[:, :, 2]

gray_image = (0.299 * R_channel + 0.587 * G_channel + 0.114 * B_channel).astype(np.uint8)

# 2. Color Image to Binary (MANUAL)

# Manual binary conversion (thresholding)
# Use the grayscale image
# If pixel > threshold: set to 255, else: set to 0
threshold_value = 127

# Method 1: Using numpy.where (simple approach)
binary_image = np.where(gray_image > threshold_value, 255, 0).astype(np.uint8)

# 3. Separate RGB Channels (MANUAL)
R_plane = rgb_image[:, :, 0]
G_plane = rgb_image[:, :, 1]
B_plane = rgb_image[:, :, 2]


# DISPLAY ALL RESULTS

fig = plt.figure(figsize=(14, 10))

# Original Color Image
plt.subplot(2, 4, 1)
plt.imshow(rgb_image)
plt.title("1. Original Color Image\n(H×W×3)", fontweight='bold')
plt.axis("off")

# Grayscale Image
plt.subplot(2, 4, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("2. Grayscale Image\n(H×W)", fontweight='bold')
plt.axis("off")

# Binary Image
plt.subplot(2, 4, 3)
plt.imshow(binary_image, cmap="gray")
plt.title("3. Binary Image\n(0 or 255)", fontweight='bold')
plt.axis("off")

# Placeholder
plt.subplot(2, 4, 4)
plt.axis("off")

# Red Plane
plt.subplot(2, 4, 5)
plt.imshow(R_plane, cmap="Reds")
plt.title("4. Red (R) Plane\n(H×W)", fontweight='bold')
plt.axis("off")

# Green Plane
plt.subplot(2, 4, 6)
plt.imshow(G_plane, cmap="Greens")
plt.title("5. Green (G) Plane\n(H×W)", fontweight='bold')
plt.axis("off")

# Blue Plane
plt.subplot(2, 4, 7)
plt.imshow(B_plane, cmap="Blues")
plt.title("6. Blue (B) Plane\n(H×W)", fontweight='bold')
plt.axis("off")

# Combined Image
plt.subplot(2, 4, 8)
plt.imshow(combined_image)
plt.title("7. Combined RGB Image\n(H×W×3)", fontweight='bold')
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read Color Image

image = cv2.imread("cat.png")

# Convert BGR to RGB
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 1. Color Image to Grayscale Image

gray_image = cv2.cvtColor(rgb_image, cv2.COLOR_RGB2GRAY)

# 2. Color Image to Binary Image
# First convert color image to grayscale
gray_for_binary = cv2.cvtColor(rgb_image, cv2.COLOR_RGB2GRAY)

# Thresholding
_, binary_image = cv2.threshold(
    gray_for_binary,
    127,
    255,
    cv2.THRESH_BINARY
)

# 3. Separate RGB Image into R, G, B Planes

R = rgb_image[:, :, 0]
G = rgb_image[:, :, 1]
B = rgb_image[:, :, 2]

# 4. Combine R, G, B Planes

# Merge the three channels
combined_image = cv2.merge([R, G, B])

# Display Results

plt.figure(figsize=(12, 10))

# Original Color Image
plt.subplot(2, 4, 1)
plt.imshow(rgb_image)
plt.title("Original Color Image")
plt.axis("off")

# Grayscale Image
plt.subplot(2, 4, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

# Binary Image
plt.subplot(2, 4, 3)
plt.imshow(binary_image, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

# Red Plane
plt.subplot(2, 4, 4)
plt.imshow(R, cmap="gray")
plt.title("Red (R) Plane")
plt.axis("off")

# Green Plane
plt.subplot(2, 4, 5)
plt.imshow(G, cmap="gray")
plt.title("Green (G) Plane")
plt.axis("off")

# Blue Plane
plt.subplot(2, 4, 6)
plt.imshow(B, cmap="gray")
plt.title("Blue (B) Plane")
plt.axis("off")

# Combined RGB Image
plt.subplot(2, 4, 7)
plt.imshow(combined_image)
plt.title("Combined RGB Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
