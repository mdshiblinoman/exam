## Problem 5
Convert a color image to grayscale and binary, separate R/G/B planes, and combine them again.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.cvtColor(cv2.imread("cat.png"), cv2.COLOR_BGR2RGB)

# Separate the image into R, G and B planes.
R = image[:, :, 0]
G = image[:, :, 1]
B = image[:, :, 2]

# Manual grayscale conversion.
gray = (0.299 * R + 0.587 * G + 0.114 * B).astype(np.uint8)

# Manual binary conversion without a thresholding function.
binary = np.zeros(gray.shape, dtype=np.uint8)
for row in range(gray.shape[0]):
    for col in range(gray.shape[1]):
        if gray[row, col] > 127:
            binary[row, col] = 255

# Combine the three planes back into an RGB image manually.
combined = np.zeros(image.shape, dtype=np.uint8)
for row in range(image.shape[0]):
    for col in range(image.shape[1]):
        combined[row, col, 0] = R[row, col]
        combined[row, col, 1] = G[row, col]
        combined[row, col, 2] = B[row, col]

plt.figure(figsize=(14, 8))
plt.subplot(2, 4, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(R, cmap="Reds")
plt.title("Red Plane")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(G, cmap="Greens")
plt.title("Green Plane")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(B, cmap="Blues")
plt.title("Blue Plane")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(combined)
plt.title("Combined RGB Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
