```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Read the image
image = cv2.imread("cat.png")
```

```python
# Manually convert BGR to RGB, then RGB to grayscale and binary
height, width, _ = image.shape
rgb_image = image.copy()
gray_image = np.zeros((height, width), dtype=np.uint8)
binary_image = np.zeros((height, width), dtype=np.uint8)

for row in range(height):
    for column in range(width):
        blue, green, red = image[row, column]
        rgb_image[row, column] = [red, green, blue]

        gray_value = int(0.299 * red + 0.587 * green + 0.114 * blue)
        gray_image[row, column] = gray_value
        binary_image[row, column] = 255 if gray_value >= 127 else 0

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(rgb_image)
plt.title("RGB Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(binary_image, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
