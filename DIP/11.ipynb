```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Image
image = cv2.imread("cat.png", 0)

# Convert to float for calculations
gray_float = image.astype(np.float32)
gray = image

# 1. Log Transformation

# Log transformation formula:
# s = c * log(1 + r)

c = 255 / np.log(1 + np.max(gray_float))

log_transform = c * np.log(1 + gray_float)

# Convert to 8-bit
log_transform = np.uint8(log_transform)

# 2. Power-Law Transformation
# Normalize pixel values between 0 and 1
normalized = gray / 255.0

# Gamma value
gamma = 0.5

# Power-law transformation:
# s = c * r^gamma

power_transform = np.power(normalized, gamma)

# Convert back to 0-255
power_transform = np.uint8(power_transform * 255)


# Display Results
plt.figure(figsize=(12, 8))

# Original Image
plt.subplot(1, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

# Log Transformation
plt.subplot(1, 3, 2)
plt.imshow(log_transform, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")

# Power-Law Transformation
plt.subplot(1, 3, 3)
plt.imshow(power_transform, cmap="gray")
plt.title(f"Power-Law (Gamma = {gamma})")
plt.axis("off")

plt.tight_layout()
plt.show()
```
