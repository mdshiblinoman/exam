```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("cat.png")

# Convert BGR to RGB
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 1. Image Negative
negative = 255 - rgb_image

# 2. Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 3. Thresholding
# Threshold value
threshold_value = 127

# Binary Thresholding
_, binary = cv2.threshold(
    gray,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)

# Display Results
plt.figure(figsize=(12, 8))

# Original Image
plt.subplot(2, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis("off")

# Negative Image
plt.subplot(2, 2, 2)
plt.imshow(negative)
plt.title("Image Negative")
plt.axis("off")

# Grayscale Image
plt.subplot(2, 2, 3)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

# Thresholded Image
plt.subplot(2, 2, 4)
plt.imshow(binary, cmap="gray")
plt.title("Binary Thresholding")
plt.axis("off")

plt.tight_layout()
plt.show()
```
