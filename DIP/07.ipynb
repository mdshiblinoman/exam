## Problem 7
Write a Python program to implement image negative and thresholding operations.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("cat.png")
height, width, channels = image.shape
gray = np.zeros((height, width), dtype=np.uint8)
negative = np.zeros((height, width), dtype=np.uint8)
binary = np.zeros((height, width), dtype=np.uint8)

threshold = 127

for row in range(height):
    for col in range(width):
        blue = int(image[row, col, 0])
        green = int(image[row, col, 1])
        red = int(image[row, col, 2])

        # Manual RGB to grayscale conversion.
        gray_value = int(0.114 * blue + 0.587 * green + 0.299 * red)
        gray[row, col] = gray_value

        # Manual negative: s = 255 - r.
        negative[row, col] = 255 - gray_value

        # Manual binary thresholding.
        if gray_value > threshold:
            binary[row, col] = 255
        else:
            binary[row, col] = 0

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image[:, :, ::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(negative, cmap="gray")
plt.title("Negative Image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(binary, cmap="gray")
plt.title("Thresholded Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
